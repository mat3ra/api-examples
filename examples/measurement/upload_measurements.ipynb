# Upload Measurements

Parse an instrument's run into the folder `mat3ra-upload` takes and upload it into a wafer's sample set on the platform. The defaults upload the Asylum Research SPM example run from mat3ra/fixtures.

Create the wafer's Sample Set in the platform under Samples before running the notebook.

## Parameters

Point `RUN_FOLDER` and `WORKFLOW_PATH` at your own run and its workflow.

```python
from pathlib import Path

import mat3ra.fixtures

HOST = "alphafilm.mat3ra.com"
WAFER_SET_NAME = "Library_1"
APPLICATION = "applications.asylum_spm.loop"
RUN_FOLDER = Path(mat3ra.fixtures.__file__).parent / "data/applications/asylum_spm/1.0/case-001"
WORKFLOW_PATH = RUN_FOLDER / "workflow.json"
UPLOAD_FOLDER = "upload"
```

## Sign in

Prints a code to confirm in the browser where you are signed in to the platform. Skip this step when already signed in.

```python
!mat3ra-auth {HOST}
```

## Parse the run

Writes the run into `UPLOAD_FOLDER`: `set.json` and one folder per measurement.

```python
import json

from mat3ra.parsers.entities.measurement_set import MeasurementSetParser

with open(WORKFLOW_PATH) as workflow_file:
    workflow = json.load(workflow_file)

MeasurementSetParser(RUN_FOLDER, workflow, format=APPLICATION).write(UPLOAD_FOLDER)
```

## Upload

Uploads the folder into the wafer's sample set. Running it again uploads only what is missing.

```python
!mat3ra-upload {WAFER_SET_NAME} {UPLOAD_FOLDER}
```
